```python
class SyntheticDataGenerator:
    """
    High-quality synthetic dataset generator for Kaggle.
    Creates realistic data with meaningful correlations and patterns.
    """
    
    def __init__(self, seed: int = 42):
        np.random.seed(seed)
        random.seed(seed)
        if fake:
            Faker.seed(seed)
    
    # ═══════════════════════════════════════════════════════════════════
    # DATASET: Employee Performance & Attrition
    # ═══════════════════════════════════════════════════════════════════
    def generate_employee_dataset(self, n_samples: int = 10000) -> pd.DataFrame:
        """
        HR dataset: attrition and performance prediction.
        Ideal for: Classification, HR analytics, churn prediction.
        """
        data = []
        departments = ['Engineering', 'Sales', 'Marketing', 'HR', 'Finance', 'Operations', 'R&D', 'Legal']
        education_levels = ['High School', 'Bachelor', 'Master', 'PhD']
        
        for i in range(n_samples):
            age = np.random.normal(38, 10)
            age = max(22, min(65, int(age)))
            
            education = np.random.choice(education_levels, p=[0.15, 0.45, 0.30, 0.10])
            education_num = education_levels.index(education) + 1
            
            department = np.random.choice(departments)
            
            # Salary correlated with age, education and department
            base_salary = 35000 + (age - 22) * 800 + education_num * 12000
            dept_multiplier = {'Engineering': 1.3, 'R&D': 1.25, 'Finance': 1.2, 'Legal': 1.15,
                              'Sales': 1.1, 'Marketing': 1.05, 'Operations': 1.0, 'HR': 0.95}
            salary = int(base_salary * dept_multiplier[department] * np.random.uniform(0.85, 1.15))
            
            years_at_company = min(age - 22, max(0, int(np.random.exponential(5))))
            
            # Performance score with multiple factors
            performance_base = 3.0
            performance_base += 0.15 * education_num
            performance_base += np.random.normal(0, 0.5)
            performance_base += 0.02 * years_at_company
            performance_score = max(1, min(5, round(performance_base, 1)))
            
            # Job satisfaction
            satisfaction = max(1, min(5, round(3 + np.random.normal(0, 1) + 
                                                0.3 * (salary / 100000) - 
                                                0.1 * abs(age - 35) / 10, 1)))
            
            work_life_balance = np.random.choice([1, 2, 3, 4], p=[0.05, 0.20, 0.50, 0.25])
            overtime = np.random.choice(['Yes', 'No'], p=[0.3, 0.7])
            
            # Distance from home
            distance_from_home = int(np.random.exponential(10)) + 1
            
            # Number of projects
            num_projects = max(1, min(10, int(np.random.poisson(4))))
            
            # Training hours
            training_hours = max(0, int(np.random.normal(40, 20)))
            
            # Attrition (target variable) - correlated with multiple factors
            attrition_prob = 0.15  # Base
            attrition_prob += 0.1 if satisfaction < 3 else -0.05
            attrition_prob += 0.15 if overtime == 'Yes' else 0
            attrition_prob += 0.05 if work_life_balance < 3 else -0.03
            attrition_prob += 0.02 * max(0, distance_from_home - 15) / 10
            attrition_prob -= 0.02 * years_at_company
            attrition_prob = max(0.02, min(0.8, attrition_prob))
            
            attrition = np.random.choice(['Yes', 'No'], p=[attrition_prob, 1-attrition_prob])
            
            data.append({
                'employee_id': f'EMP{i+1:05d}',
                'age': age,
                'gender': np.random.choice(['Male', 'Female', 'Other'], p=[0.48, 0.48, 0.04]),
                'education': education,
                'department': department,
                'job_role': fake.job() if fake else f'Role_{np.random.randint(1, 50)}',
                'monthly_salary': salary,
                'years_at_company': years_at_company,
                'years_in_current_role': min(years_at_company, max(0, int(np.random.exponential(3)))),
                'performance_score': performance_score,
                'job_satisfaction': satisfaction,
                'work_life_balance': work_life_balance,
                'overtime': overtime,
                'distance_from_home_km': distance_from_home,
                'num_projects_last_year': num_projects,
                'training_hours_last_year': training_hours,
                'num_promotions': min(years_at_company // 3, np.random.poisson(1)),
                'last_promotion_years_ago': min(years_at_company, int(np.random.exponential(2))),
                'stock_option_level': np.random.choice([0, 1, 2, 3], p=[0.4, 0.3, 0.2, 0.1]),
                'relationship_satisfaction': np.random.choice([1, 2, 3, 4], p=[0.1, 0.2, 0.4, 0.3]),
                'environment_satisfaction': np.random.choice([1, 2, 3, 4], p=[0.1, 0.2, 0.4, 0.3]),
                'attrition': attrition
            })
        
        return pd.DataFrame(data)
    
    # ═══════════════════════════════════════════════════════════════════
    # DATASET: Healthcare - Disease Prediction
    # ═══════════════════════════════════════════════════════════════════
    def generate_healthcare_dataset(self, n_samples: int = 15000) -> pd.DataFrame:
        """
        Medical dataset: cardiovascular disease prediction.
        Ideal for: Binary classification, medical analysis.
        """
        data = []
        
        for i in range(n_samples):
            age = int(np.random.normal(50, 15))
            age = max(18, min(90, age))
            
            gender = np.random.choice(['Male', 'Female'], p=[0.48, 0.52])
            
            # BMI correlated with age
            bmi_base = 22 + (age - 30) * 0.1
            bmi = max(15, min(50, round(np.random.normal(bmi_base, 5), 1)))
            
            # Blood pressure correlated with age and BMI
            systolic_base = 110 + (age - 30) * 0.5 + (bmi - 25) * 1.5
            systolic = int(max(90, min(200, np.random.normal(systolic_base, 15))))
            diastolic = int(systolic * np.random.uniform(0.55, 0.70))
            
            # Cholesterol
            cholesterol_base = 180 + (age - 30) * 1 + (bmi - 25) * 3
            cholesterol = int(max(120, min(350, np.random.normal(cholesterol_base, 30))))
            
            # Glucose
            glucose_base = 90 + (age - 30) * 0.3 + (bmi - 25) * 2
            glucose = int(max(60, min(250, np.random.normal(glucose_base, 20))))
            
            # Lifestyle factors
            smoking = np.random.choice(['Never', 'Former', 'Current'], 
                                       p=[0.55, 0.25, 0.20])
            alcohol = np.random.choice(['None', 'Moderate', 'Heavy'],
                                       p=[0.30, 0.55, 0.15])
            physical_activity = np.random.choice(['Low', 'Moderate', 'High'],
                                                  p=[0.25, 0.50, 0.25])
            
            # Family history
            family_history = np.random.choice([0, 1], p=[0.70, 0.30])
            
            # Sleep hours
            sleep_hours = max(3, min(12, round(np.random.normal(7, 1.5), 1)))
            
            # Stress
            stress_level = np.random.choice(['Low', 'Medium', 'High'], p=[0.30, 0.45, 0.25])
            
            # Cardiovascular risk calculation (target variable)
            risk_score = 0
            risk_score += (age - 40) * 0.02 if age > 40 else 0
            risk_score += 0.15 if gender == 'Male' else 0
            risk_score += (bmi - 25) * 0.03 if bmi > 25 else 0
            risk_score += (systolic - 120) * 0.01 if systolic > 120 else 0
            risk_score += (cholesterol - 200) * 0.005 if cholesterol > 200 else 0
            risk_score += (glucose - 100) * 0.005 if glucose > 100 else 0
            risk_score += 0.2 if smoking == 'Current' else (0.1 if smoking == 'Former' else 0)
            risk_score += 0.1 if alcohol == 'Heavy' else 0
            risk_score -= 0.1 if physical_activity == 'High' else (0.05 if physical_activity == 'Moderate' else 0)
            risk_score += 0.15 if family_history else 0
            risk_score += 0.05 if stress_level == 'High' else 0
            risk_score -= 0.05 if 6 <= sleep_hours <= 8 else 0
            
            risk_prob = 1 / (1 + np.exp(-risk_score))  # Sigmoid
            heart_disease = np.random.choice([0, 1], p=[1-risk_prob, risk_prob])
            
            data.append({
                'patient_id': f'PAT{i+1:06d}',
                'age': age,
                'gender': gender,
                'height_cm': int(np.random.normal(170 if gender == 'Male' else 162, 8)),
                'weight_kg': round(bmi * ((170 if gender == 'Male' else 162) / 100) ** 2, 1),
                'bmi': bmi,
                'systolic_bp': systolic,
                'diastolic_bp': diastolic,
                'cholesterol_total': cholesterol,
                'cholesterol_hdl': int(max(20, min(100, np.random.normal(50, 15)))),
                'cholesterol_ldl': int(max(50, min(200, cholesterol * 0.6 + np.random.normal(0, 20)))),
                'triglycerides': int(max(50, min(400, np.random.normal(150, 50)))),
                'glucose_fasting': glucose,
                'hba1c': round(max(4, min(12, glucose / 25 + np.random.normal(0, 0.5))), 1),
                'smoking_status': smoking,
                'alcohol_consumption': alcohol,
                'physical_activity': physical_activity,
                'sleep_hours_avg': sleep_hours,
                'stress_level': stress_level,
                'family_history_heart_disease': family_history,
                'diabetes': 1 if glucose > 126 or np.random.random() < 0.1 else 0,
                'hypertension': 1 if systolic > 140 or diastolic > 90 else 0,
                'heart_disease': heart_disease
            })
        
        return pd.DataFrame(data)
    
    # ═══════════════════════════════════════════════════════════════════
    # DATASET: Gaming - Player Behavior & Monetization
    # ═══════════════════════════════════════════════════════════════════
    def generate_gaming_dataset(self, n_samples: int = 20000) -> pd.DataFrame:
        """
        Gaming dataset: player behavior and monetization.
        Ideal for: Churn prediction, LTV prediction, segmentation.
        """
        data = []
        platforms = ['PC', 'PlayStation', 'Xbox', 'Mobile', 'Nintendo Switch']
        genres = ['FPS', 'RPG', 'MOBA', 'Battle Royale', 'Sports', 'Racing', 'Strategy', 'Casual']
        countries = ['USA', 'UK', 'Germany', 'France', 'Japan', 'Brazil', 'South Korea', 'Canada', 'Australia', 'Spain']
        
        for i in range(n_samples):
            # Demographics
            age = int(np.random.normal(28, 10))
            age = max(13, min(65, age))
            
            platform = np.random.choice(platforms, p=[0.30, 0.20, 0.15, 0.25, 0.10])
            
            # Play time based on age and platform
            if platform == 'Mobile':
                daily_play_hours = max(0.1, min(8, np.random.exponential(1.5)))
            else:
                daily_play_hours = max(0.5, min(12, np.random.exponential(2.5)))
            
            # Age adjustment
            if age > 35:
                daily_play_hours *= 0.7
            elif age < 20:
                daily_play_hours *= 1.2
            
            daily_play_hours = round(daily_play_hours, 1)
            
            # Days since registration
            days_since_registration = int(np.random.exponential(180)) + 1
            
            # Sessions
            sessions_per_week = max(1, min(50, int(np.random.poisson(7 * daily_play_hours / 2))))
            
            # Player level
            level = min(100, int(days_since_registration * daily_play_hours * 0.1) + np.random.randint(1, 10))
            
            # Achievements
            achievements_unlocked = min(200, int(level * 1.5 + np.random.poisson(10)))
            
            # Social features
            friends_count = int(np.random.exponential(15))
            guild_member = np.random.choice([0, 1], p=[0.6, 0.4])
            
            # Spending behavior
            is_spender = np.random.choice([0, 1], p=[0.75, 0.25])
            
            if is_spender:
                # Whale detection logic
                whale_prob = 0.05 if daily_play_hours > 3 else 0.01
                is_whale = np.random.choice([0, 1], p=[1-whale_prob, whale_prob])
                
                if is_whale:
                    total_spent = round(np.random.exponential(500) + 200, 2)
                else:
                    total_spent = round(np.random.exponential(30) + 5, 2)
                
                purchases_count = max(1, int(total_spent / np.random.uniform(5, 25)))
            else:
                total_spent = 0
                purchases_count = 0
                is_whale = 0
            
            # Churn probability
            churn_base = 0.3
            churn_base -= 0.1 if daily_play_hours > 2 else 0
            churn_base -= 0.15 if total_spent > 50 else (-0.05 if total_spent > 0 else 0)
            churn_base -= 0.1 if guild_member else 0
            churn_base -= 0.05 if friends_count > 10 else 0
            churn_base += 0.2 if days_since_registration < 7 else 0
            churn_base = max(0.05, min(0.8, churn_base))
            
            churned = np.random.choice([0, 1], p=[1-churn_base, churn_base])
            
            # Customer lifetime value (for non-churned users)
            if not churned:
                predicted_ltv = total_spent * np.random.uniform(1.5, 4) + daily_play_hours * 10
            else:
                predicted_ltv = total_spent
            
            data.append({
                'player_id': f'PLR{i+1:06d}',
                'age': age,
                'gender': np.random.choice(['Male', 'Female', 'Other'], p=[0.65, 0.30, 0.05]),
                'country': np.random.choice(countries),
                'platform': platform,
                'favorite_genre': np.random.choice(genres),
                'days_since_registration': days_since_registration,
                'daily_play_hours_avg': daily_play_hours,
                'sessions_per_week': sessions_per_week,
                'avg_session_duration_min': round(daily_play_hours * 60 / max(1, sessions_per_week / 7), 1),
                'level': level,
                'achievements_unlocked': achievements_unlocked,
                'total_matches_played': int(days_since_registration * daily_play_hours * 2),
                'win_rate': round(np.random.beta(5, 5), 3),
                'friends_count': friends_count,
                'guild_member': guild_member,
                'messages_sent': int(np.random.exponential(50)) if friends_count > 0 else 0,
                'is_premium_subscriber': np.random.choice([0, 1], p=[0.85, 0.15]),
                'total_spent_usd': round(total_spent, 2),
                'purchases_count': purchases_count,
                'days_since_last_purchase': np.random.randint(0, 90) if purchases_count > 0 else -1,
                'ads_watched': int(np.random.exponential(100)) if platform == 'Mobile' and total_spent < 10 else 0,
                'support_tickets': np.random.poisson(0.5),
                'reported_bugs': np.random.poisson(0.2),
                'is_whale': is_whale,
                'predicted_ltv': round(predicted_ltv, 2),
                'churned': churned
            })
        
        return pd.DataFrame(data)
    
    # ═══════════════════════════════════════════════════════════════════
    # DATASET: Real Estate - Price Prediction
    # ═══════════════════════════════════════════════════════════════════
    def generate_real_estate_dataset(self, n_samples: int = 12000) -> pd.DataFrame:
        """
        Real estate dataset: house price prediction.
        Ideal for: Regression, market analysis.
        """
        data = []
        cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 
                  'Philadelphia', 'San Antonio', 'San Diego', 'Dallas', 'Austin',
                  'Miami', 'Seattle', 'Denver', 'Boston', 'Portland']
        
        property_types = ['Apartment', 'House', 'Condo', 'Townhouse', 'Studio']
        
        # Base prices by city (per sqft)
        city_price_base = {
            'New York': 800, 'Los Angeles': 650, 'San Diego': 550, 'Boston': 600,
            'Seattle': 500, 'Miami': 400, 'Chicago': 300, 'Denver': 400,
            'Austin': 350, 'Portland': 380, 'Dallas': 280, 'Houston': 250,
            'Phoenix': 280, 'Philadelphia': 320, 'San Antonio': 220
        }
        
        for i in range(n_samples):
            city = np.random.choice(cities)
            property_type = np.random.choice(property_types, p=[0.25, 0.35, 0.20, 0.15, 0.05])
            
            # Size based on type
            if property_type == 'Studio':
                sqft = int(np.random.normal(450, 100))
            elif property_type == 'Apartment':
                sqft = int(np.random.normal(900, 300))
            elif property_type == 'Condo':
                sqft = int(np.random.normal(1200, 400))
            elif property_type == 'Townhouse':
                sqft = int(np.random.normal(1800, 500))
            else:  # House
                sqft = int(np.random.normal(2200, 700))
            
            sqft = max(300, min(6000, sqft))
            
            # Bedrooms based on size
            bedrooms = max(0, min(6, int(sqft / 500)))
            bathrooms = max(1, min(5, int(bedrooms * 0.8 + np.random.choice([0, 0.5, 1]))))
            
            # Year built
            year_built = int(np.random.normal(1985, 25))
            year_built = max(1900, min(2024, year_built))
            
            # Condition based on age and maintenance
            age = 2024 - year_built
            condition_score = max(1, min(10, 10 - age/15 + np.random.normal(0, 1.5)))
            condition_score = round(condition_score, 1)
            
            # Features
            has_garage = np.random.choice([0, 1], p=[0.3, 0.7] if property_type in ['House', 'Townhouse'] else [0.7, 0.3])
            has_pool = np.random.choice([0, 1], p=[0.85, 0.15]) if property_type == 'House' else 0
            has_garden = np.random.choice([0, 1], p=[0.4, 0.6]) if property_type in ['House', 'Townhouse'] else 0
            
            # Location
            distance_to_center = max(0.5, np.random.exponential(8))
            crime_rate = max(0, min(100, np.random.normal(30, 15) + distance_to_center))
            school_rating = max(1, min(10, 10 - crime_rate/20 + np.random.normal(0, 1)))
            
            # Parking spots
            parking_spots = 0 if property_type == 'Studio' else (2 if has_garage else np.random.choice([0, 1]))
            
            # Price calculation
            base_price = city_price_base[city]
            price_per_sqft = base_price
            price_per_sqft *= (1 + (condition_score - 5) * 0.05)  # Condition
            price_per_sqft *= (1 - distance_to_center * 0.015)  # Distance to center
            price_per_sqft *= (1 + school_rating * 0.02)  # School rating
            price_per_sqft *= np.random.uniform(0.9, 1.1)  # Random variation
            
            price = sqft * price_per_sqft
            price += has_garage * 25000
            price += has_pool * 35000
            price += has_garden * 15000
            price += parking_spots * 10000
            
            price = int(max(50000, price))
            
            # HOA fees
            hoa_fee = 0
            if property_type in ['Apartment', 'Condo']:
                hoa_fee = int(np.random.normal(300, 100))
                hoa_fee = max(100, min(800, hoa_fee))
            
            data.append({
                'property_id': f'PROP{i+1:06d}',
                'city': city,
                'neighborhood': fake.street_name() if fake else f'Neighborhood_{np.random.randint(1, 100)}',
                'property_type': property_type,
                'sqft': sqft,
                'lot_size_sqft': sqft * np.random.uniform(1.5, 4) if property_type == 'House' else sqft,
                'bedrooms': bedrooms,
                'bathrooms': bathrooms,
                'floors': np.random.choice([1, 2, 3], p=[0.4, 0.45, 0.15]) if property_type in ['House', 'Townhouse'] else 1,
                'year_built': year_built,
                'year_renovated': year_built + np.random.randint(10, 30) if np.random.random() < 0.3 else None,
                'condition_score': condition_score,
                'has_garage': has_garage,
                'parking_spots': parking_spots,
                'has_pool': has_pool,
                'has_garden': has_garden,
                'has_basement': np.random.choice([0, 1], p=[0.6, 0.4]) if property_type in ['House', 'Townhouse'] else 0,
                'heating_type': np.random.choice(['Gas', 'Electric', 'Oil', 'Solar'], p=[0.5, 0.35, 0.10, 0.05]),
                'cooling_type': np.random.choice(['Central', 'Window', 'None'], p=[0.6, 0.25, 0.15]),
                'distance_to_center_miles': round(distance_to_center, 1),
                'distance_to_public_transport_miles': round(max(0.1, np.random.exponential(1)), 1),
                'school_rating': round(school_rating, 1),
                'crime_rate_index': round(crime_rate, 1),
                'walk_score': int(max(0, min(100, 100 - distance_to_center * 5 + np.random.normal(0, 10)))),
                'hoa_fee_monthly': hoa_fee,
                'property_tax_annual': int(price * np.random.uniform(0.008, 0.025)),
                'days_on_market': int(np.random.exponential(30)) + 1,
                'price': price
            })
        
        return pd.DataFrame(data)
    
    # ═══════════════════════════════════════════════════════════════════
    # DATASET: E-commerce - Customer Behavior
    # ═══════════════════════════════════════════════════════════════════
    def generate_ecommerce_dataset(self, n_samples: int = 25000) -> pd.DataFrame:
        """
        E-commerce dataset: customer purchase behavior.
        Ideal for: Segmentation, recommendation, purchase prediction.
        """
        data = []
        categories = ['Electronics', 'Fashion', 'Home & Garden', 'Sports', 'Books', 
                     'Beauty', 'Toys', 'Food & Grocery', 'Automotive', 'Health']
        devices = ['Desktop', 'Mobile', 'Tablet']
        traffic_sources = ['Organic Search', 'Paid Search', 'Social Media', 'Email', 'Direct', 'Referral']
        
        for i in range(n_samples):
            # Demographics
            age = int(np.random.normal(35, 12))
            age = max(18, min(75, age))
            
            gender = np.random.choice(['Male', 'Female', 'Other'], p=[0.45, 0.50, 0.05])
            
            # Browsing behavior
            device = np.random.choice(devices, p=[0.35, 0.55, 0.10])
            traffic_source = np.random.choice(traffic_sources, p=[0.25, 0.20, 0.20, 0.15, 0.12, 0.08])
            
            # Time on site
            if device == 'Mobile':
                session_duration = max(10, np.random.exponential(180))
            else:
                session_duration = max(30, np.random.exponential(300))
            
            pages_viewed = max(1, int(session_duration / 60 * np.random.uniform(0.8, 1.5)))
            
            # Products
            products_viewed = max(1, int(pages_viewed * 0.6))
            products_added_to_cart = int(products_viewed * np.random.beta(2, 5))
            
            # Conversion
            conversion_base = 0.03
            conversion_base += 0.02 if products_added_to_cart > 0 else 0
            conversion_base += 0.01 if session_duration > 300 else 0
            conversion_base += 0.02 if traffic_source == 'Email' else 0
            conversion_base += 0.01 if device == 'Desktop' else 0
            
            converted = np.random.choice([0, 1], p=[1-conversion_base, conversion_base])
            
            # Order value if converted
            if converted:
                order_value = np.random.exponential(80) + 20
                order_value = round(min(500, order_value), 2)
                items_purchased = max(1, int(order_value / 40))
            else:
                order_value = 0
                items_purchased = 0
            
            # Customer history
            is_returning = np.random.choice([0, 1], p=[0.6, 0.4])
            
            if is_returning:
                previous_orders = int(np.random.exponential(3)) + 1
                days_since_last_visit = int(np.random.exponential(30))
                customer_lifetime_value = round(previous_orders * np.random.uniform(50, 150) + order_value, 2)
            else:
                previous_orders = 0
                days_since_last_visit = -1
                customer_lifetime_value = order_value
            
            # Engagement
            newsletter_subscriber = np.random.choice([0, 1], p=[0.7, 0.3])
            reviews_written = int(np.random.exponential(0.5)) if previous_orders > 0 else 0
            
            # Favorite category
            favorite_category = np.random.choice(categories)
            
            data.append({
                'session_id': f'SES{i+1:07d}',
                'customer_id': f'CUST{np.random.randint(1, n_samples//3):06d}' if is_returning else f'CUST{n_samples//3 + i:06d}',
                'age': age,
                'gender': gender,
                'country': np.random.choice(['USA', 'UK', 'Germany', 'France', 'Canada', 'Australia', 'Spain', 'Italy', 'Japan', 'Brazil']),
                'device': device,
                'browser': np.random.choice(['Chrome', 'Safari', 'Firefox', 'Edge', 'Other'], p=[0.55, 0.20, 0.12, 0.08, 0.05]),
                'traffic_source': traffic_source,
                'landing_page': np.random.choice(['Homepage', 'Product Page', 'Category Page', 'Search Results', 'Promo Page'], p=[0.30, 0.25, 0.20, 0.15, 0.10]),
                'session_duration_seconds': int(session_duration),
                'pages_viewed': pages_viewed,
                'products_viewed': products_viewed,
                'products_added_to_cart': products_added_to_cart,
                'cart_abandonment': 1 if products_added_to_cart > 0 and not converted else 0,
                'used_search': np.random.choice([0, 1], p=[0.4, 0.6]),
                'used_filters': np.random.choice([0, 1], p=[0.5, 0.5]),
                'clicked_recommendation': np.random.choice([0, 1], p=[0.7, 0.3]),
                'favorite_category': favorite_category,
                'is_returning_customer': is_returning,
                'previous_orders': previous_orders,
                'days_since_last_visit': days_since_last_visit,
                'newsletter_subscriber': newsletter_subscriber,
                'has_account': 1 if is_returning or np.random.random() < 0.3 else 0,
                'used_coupon': np.random.choice([0, 1], p=[0.8, 0.2]) if converted else 0,
                'coupon_discount_percent': np.random.choice([5, 10, 15, 20, 25]) if np.random.random() < 0.2 else 0,
                'reviews_written': reviews_written,
                'wishlist_items': int(np.random.exponential(2)),
                'converted': converted,
                'order_value': order_value,
                'items_purchased': items_purchased,
                'customer_lifetime_value': customer_lifetime_value
            })
        
        return pd.DataFrame(data)
    
    # ═══════════════════════════════════════════════════════════════════
    # DATASET: Used Cars - Price Prediction
    # ═══════════════════════════════════════════════════════════════════
    def generate_used_cars_dataset(self, n_samples: int = 15000) -> pd.DataFrame:
        """
        Used cars dataset: price prediction.
        Ideal for: Regression, automotive market analysis.
        """
        data = []
        
        brands_models = {
            'Toyota': ['Camry', 'Corolla', 'RAV4', 'Highlander', 'Tacoma', 'Prius'],
            'Honda': ['Civic', 'Accord', 'CR-V', 'Pilot', 'HR-V'],
            'Ford': ['F-150', 'Mustang', 'Explorer', 'Escape', 'Focus'],
            'Chevrolet': ['Silverado', 'Malibu', 'Equinox', 'Tahoe', 'Camaro'],
            'BMW': ['3 Series', '5 Series', 'X3', 'X5', 'M3'],
            'Mercedes-Benz': ['C-Class', 'E-Class', 'GLC', 'GLE', 'S-Class'],
            'Audi': ['A4', 'A6', 'Q5', 'Q7', 'A3'],
            'Nissan': ['Altima', 'Sentra', 'Rogue', 'Pathfinder', 'Maxima'],
            'Hyundai': ['Elantra', 'Sonata', 'Tucson', 'Santa Fe', 'Kona'],
            'Volkswagen': ['Jetta', 'Passat', 'Tiguan', 'Atlas', 'Golf']
        }
        
        # Approximate MSRP by brand (multiplier factor)
        brand_msrp_factor = {
            'Toyota': 1.0, 'Honda': 1.0, 'Nissan': 0.95, 'Hyundai': 0.85,
            'Ford': 0.95, 'Chevrolet': 0.95, 'Volkswagen': 1.05,
            'BMW': 1.8, 'Mercedes-Benz': 1.9, 'Audi': 1.7
        }
        
        fuel_types = ['Gasoline', 'Diesel', 'Hybrid', 'Electric']
        transmissions = ['Automatic', 'Manual', 'CVT']
        colors = ['White', 'Black', 'Silver', 'Gray', 'Blue', 'Red', 'Green', 'Brown', 'Beige', 'Other']
        
        for i in range(n_samples):
            brand = np.random.choice(list(brands_models.keys()))
            model = np.random.choice(brands_models[brand])
            
            # Car year
            year = np.random.randint(2005, 2025)
            age = 2025 - year
            
            # Mileage based on age
            mileage_per_year = np.random.normal(12000, 4000)
            mileage = int(max(1000, age * mileage_per_year * np.random.uniform(0.7, 1.3)))
            
            # Fuel type
            if brand in ['Toyota', 'Honda'] and np.random.random() < 0.15:
                fuel_type = 'Hybrid'
            elif year >= 2018 and np.random.random() < 0.08:
                fuel_type = 'Electric'
            else:
                fuel_type = np.random.choice(['Gasoline', 'Diesel'], p=[0.85, 0.15])
            
            # Transmission
            if brand in ['BMW', 'Mercedes-Benz', 'Audi']:
                transmission = np.random.choice(transmissions, p=[0.85, 0.10, 0.05])
            else:
                transmission = np.random.choice(transmissions, p=[0.70, 0.15, 0.15])
            
            # Engine
            if fuel_type == 'Electric':
                engine_size = 0
                horsepower = np.random.randint(200, 450)
            else:
                engine_size = np.random.choice([1.5, 1.8, 2.0, 2.4, 2.5, 3.0, 3.5, 4.0, 5.0],
                                               p=[0.10, 0.15, 0.20, 0.15, 0.15, 0.10, 0.08, 0.05, 0.02])
                horsepower = int(engine_size * np.random.uniform(60, 90))
            
            # Number of owners
            owners = min(5, max(1, int(np.random.exponential(1.2)) + 1))
            
            # Accidents
            accidents = int(np.random.exponential(0.3))
            
            # Condition
            condition_base = 5 - age * 0.15 - accidents * 0.5 - (owners - 1) * 0.2
            condition = max(1, min(5, round(condition_base + np.random.normal(0, 0.5), 1)))
            
            # Features
            has_sunroof = np.random.choice([0, 1], p=[0.7, 0.3])
            has_leather = np.random.choice([0, 1], p=[0.6, 0.4] if brand in ['BMW', 'Mercedes-Benz', 'Audi'] else [0.8, 0.2])
            has_navigation = np.random.choice([0, 1], p=[0.5, 0.5] if year >= 2015 else [0.8, 0.2])
            has_backup_camera = np.random.choice([0, 1], p=[0.3, 0.7] if year >= 2015 else [0.7, 0.3])
            bluetooth = np.random.choice([0, 1], p=[0.2, 0.8] if year >= 2012 else [0.6, 0.4])
            
            # Price calculation
            base_price = 28000
            base_price *= brand_msrp_factor[brand]
            
            # Depreciation by age
            depreciation = 0.15 * min(age, 1) + 0.10 * min(max(0, age-1), 2) + 0.05 * max(0, age-3)
            price = base_price * (1 - depreciation)
            
            # Adjustments
            price *= (1 - mileage / 500000)  # Mileage depreciation
            price *= (0.9 + condition * 0.04)  # Condition
            price -= accidents * 2000  # Accidents
            price += has_sunroof * 1000
            price += has_leather * 1500
            price += has_navigation * 800
            price += (horsepower - 150) * 20  # Power
            price *= np.random.uniform(0.92, 1.08)  # Market variation
            
            price = int(max(2000, price))
            
            data.append({
                'listing_id': f'CAR{i+1:06d}',
                'brand': brand,
                'model': model,
                'year': year,
                'mileage_km': mileage,
                'fuel_type': fuel_type,
                'transmission': transmission,
                'engine_size_l': engine_size,
                'horsepower': horsepower,
                'drivetrain': np.random.choice(['FWD', 'RWD', 'AWD'], p=[0.50, 0.20, 0.30]),
                'exterior_color': np.random.choice(colors),
                'interior_color': np.random.choice(['Black', 'Gray', 'Beige', 'Brown', 'White'], p=[0.40, 0.25, 0.20, 0.10, 0.05]),
                'num_owners': owners,
                'accidents_reported': accidents,
                'condition_score': condition,
                'title_status': np.random.choice(['Clean', 'Salvage', 'Rebuilt'], p=[0.92, 0.05, 0.03]),
                'has_sunroof': has_sunroof,
                'has_leather_seats': has_leather,
                'has_navigation': has_navigation,
                'has_backup_camera': has_backup_camera,
                'has_bluetooth': bluetooth,
                'has_heated_seats': np.random.choice([0, 1], p=[0.6, 0.4]),
                'has_cruise_control': np.random.choice([0, 1], p=[0.3, 0.7]),
                'mpg_city': int(30 - engine_size * 3 + np.random.normal(0, 2)) if fuel_type != 'Electric' else 0,
                'mpg_highway': int(38 - engine_size * 3 + np.random.normal(0, 2)) if fuel_type != 'Electric' else 0,
                'warranty_remaining_months': max(0, (year - 2020) * 12 + np.random.randint(-24, 24)) if year >= 2020 else 0,
                'service_records_available': np.random.choice([0, 1], p=[0.3, 0.7]),
                'location_state': np.random.choice(['CA', 'TX', 'FL', 'NY', 'IL', 'PA', 'OH', 'GA', 'NC', 'MI']),
                'days_on_market': int(np.random.exponential(25)) + 1,
                'price_usd': price
            })
        
        return pd.DataFrame(data)

# Create generator instance
generator = SyntheticDataGenerator(seed=42)
print("[OK] SyntheticDataGenerator class loaded")
```

## 2. Web Scraping - Real Data Extraction

Tools to extract data from public sources and create unique datasets.


```python
class WebScraper:
    """
    Web scraping tools to create unique datasets.
    Always respect robots.txt and terms of service.
    """
    
    def __init__(self):
        self.headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
            'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8',
            'Accept-Language': 'en-US,en;q=0.5',
        }
        self.session = requests.Session()
        self.session.headers.update(self.headers)
    
    def _safe_request(self, url: str, delay: float = 1.0) -> Optional[BeautifulSoup]:
        """Performs request with delay to be respectful to the server."""
        import time
        time.sleep(delay)
        try:
            response = self.session.get(url, timeout=15)
            response.raise_for_status()
            return BeautifulSoup(response.content, 'html.parser')
        except Exception as e:
            print(f"Request error: {e}")
            return None
    
    # ═══════════════════════════════════════════════════════════════════
    # SCRAPER: Wikipedia - List Extraction
    # ═══════════════════════════════════════════════════════════════════
    def scrape_wikipedia_table(self, url: str, table_index: int = 0) -> pd.DataFrame:
        """
        Extracts tables from Wikipedia.
        Example: List of countries, cities, movies, etc.
        """
        soup = self._safe_request(url)
        if soup is None:
            return pd.DataFrame()
        
        tables = soup.find_all('table', {'class': 'wikitable'})
        if not tables or table_index >= len(tables):
            print(f"Table not found at index {table_index}")
            return pd.DataFrame()
        
        # Usar pandas para parsear la tabla HTML
        html_table = str(tables[table_index])
        df = pd.read_html(html_table)[0]
        
        return df
    
    # ═══════════════════════════════════════════════════════════════════
    # SCRAPER: GitHub Trending Repositories  
    # ═══════════════════════════════════════════════════════════════════
    def scrape_github_trending(self, language: str = '', since: str = 'daily') -> pd.DataFrame:
        """
        Extracts trending repositories from GitHub.
        language: python, javascript, etc. (empty for all)
        since: daily, weekly, monthly
        """
        url = f"https://github.com/trending/{language}?since={since}"
        soup = self._safe_request(url)
        if soup is None:
            return pd.DataFrame()
        
        repos = []
        articles = soup.find_all('article', {'class': 'Box-row'})
        
        for article in articles:
            try:
                # Repo name
                h2 = article.find('h2')
                if h2:
                    repo_link = h2.find('a')
                    repo_name = repo_link.text.strip().replace('\n', '').replace(' ', '')
                    repo_url = 'https://github.com' + repo_link['href']
                else:
                    continue
                
                # Description
                desc_p = article.find('p')
                description = desc_p.text.strip() if desc_p else ''
                
                # Language
                lang_span = article.find('span', {'itemprop': 'programmingLanguage'})
                prog_language = lang_span.text.strip() if lang_span else 'Unknown'
                
                # Stars
                stars_a = article.find('a', href=lambda x: x and '/stargazers' in x)
                stars = stars_a.text.strip().replace(',', '') if stars_a else '0'
                
                # Forks
                forks_a = article.find('a', href=lambda x: x and '/forks' in x)
                forks = forks_a.text.strip().replace(',', '') if forks_a else '0'
                
                # Stars today
                stars_today_span = article.find('span', {'class': 'd-inline-block float-sm-right'})
                stars_today = '0'
                if stars_today_span:
                    stars_today_text = stars_today_span.text.strip()
                    stars_today = ''.join(filter(str.isdigit, stars_today_text.split()[0]))
                
                repos.append({
                    'repository': repo_name,
                    'url': repo_url,
                    'description': description[:200] if description else '',
                    'language': prog_language,
                    'total_stars': int(stars) if stars else 0,
                    'total_forks': int(forks) if forks else 0,
                    'stars_today': int(stars_today) if stars_today else 0,
                    'scraped_date': datetime.now().strftime('%Y-%m-%d'),
                    'time_range': since
                })
            except Exception as e:
                continue
        
        return pd.DataFrame(repos)
    
    # ═══════════════════════════════════════════════════════════════════
    # SCRAPER: IMDb Top Movies
    # ═══════════════════════════════════════════════════════════════════
    def scrape_imdb_top_movies(self, chart: str = 'top') -> pd.DataFrame:
        """
        Extracts movie information from IMDb.
        chart: 'top' for Top 250, 'boxoffice' for box office
        """
        if chart == 'top':
            url = "https://www.imdb.com/chart/top/"
        else:
            url = "https://www.imdb.com/chart/boxoffice/"
        
        soup = self._safe_request(url)
        if soup is None:
            return pd.DataFrame()
        
        movies = []
        
        # IMDb uses different structures, we try both
        try:
            # Find list items
            items = soup.find_all('li', {'class': lambda x: x and 'ipc-metadata-list-summary-item' in str(x)})
            
            for i, item in enumerate(items[:250], 1):
                try:
                    # Title
                    title_elem = item.find('h3')
                    title = title_elem.text.strip() if title_elem else ''
                    # Clean number from title
                    if title and title[0].isdigit():
                        title = '.'.join(title.split('.')[1:]).strip()
                    
                    # Year, duration, rating
                    metadata = item.find_all('span', {'class': lambda x: x and 'cli-title-metadata-item' in str(x)})
                    year = metadata[0].text.strip() if len(metadata) > 0 else ''
                    duration = metadata[1].text.strip() if len(metadata) > 1 else ''
                    age_rating = metadata[2].text.strip() if len(metadata) > 2 else ''
                    
                    # IMDb Rating
                    rating_elem = item.find('span', {'class': lambda x: x and 'ipc-rating-star' in str(x)})
                    rating = ''
                    if rating_elem:
                        rating_text = rating_elem.text.strip()
                        rating = rating_text.split()[0] if rating_text else ''
                    
                    movies.append({
                        'rank': i,
                        'title': title,
                        'year': year,
                        'duration': duration,
                        'age_rating': age_rating,
                        'imdb_rating': float(rating) if rating else None,
                        'scraped_date': datetime.now().strftime('%Y-%m-%d')
                    })
                except Exception:
                    continue
        except Exception as e:
            print(f"Error parsing IMDb: {e}")
        
        return pd.DataFrame(movies)
    
    # ═══════════════════════════════════════════════════════════════════
    # SCRAPER: Books from Open Library
    # ═══════════════════════════════════════════════════════════════════
    def scrape_books_openlibrary(self, subject: str = 'science_fiction', limit: int = 100) -> pd.DataFrame:
        """
        Extracts books from Open Library API (free and legal).
        """
        url = f"https://openlibrary.org/subjects/{subject}.json?limit={limit}"
        
        try:
            response = self.session.get(url, timeout=15)
            response.raise_for_status()
            data = response.json()
            
            books = []
            for work in data.get('works', []):
                books.append({
                    'title': work.get('title', ''),
                    'authors': ', '.join([a.get('name', '') for a in work.get('authors', [])]),
                    'first_publish_year': work.get('first_publish_year', ''),
                    'subject': subject,
                    'cover_id': work.get('cover_id', ''),
                    'edition_count': work.get('edition_count', 0),
                    'has_fulltext': work.get('has_fulltext', False),
                    'key': work.get('key', '')
                })
            
            return pd.DataFrame(books)
        except Exception as e:
            print(f"Error: {e}")
            return pd.DataFrame()
    
    # ═══════════════════════════════════════════════════════════════════
    # SCRAPER: Music Charts (Spotify Public Data)
    # ═══════════════════════════════════════════════════════════════════
    def scrape_music_kworb(self) -> pd.DataFrame:
        """
        Extracts music chart data from kworb.net (public Spotify data).
        """
        url = "https://kworb.net/spotify/artists.html"
        soup = self._safe_request(url)
        if soup is None:
            return pd.DataFrame()
        
        artists = []
        table = soup.find('table')
        if table:
            rows = table.find_all('tr')[1:]  # Skip header
            for row in rows[:200]:  # Top 200
                cols = row.find_all('td')
                if len(cols) >= 4:
                    artists.append({
                        'rank': len(artists) + 1,
                        'artist': cols[0].text.strip(),
                        'monthly_listeners': cols[1].text.strip().replace(',', ''),
                        'daily_trend': cols[2].text.strip(),
                        'peak_listeners': cols[3].text.strip().replace(',', '') if len(cols) > 3 else '',
                        'scraped_date': datetime.now().strftime('%Y-%m-%d')
                    })
        
        return pd.DataFrame(artists)
    
    # ═══════════════════════════════════════════════════════════════════
    # SCRAPER: Sports Statistics
    # ═══════════════════════════════════════════════════════════════════
    def scrape_nba_stats_reference(self) -> pd.DataFrame:
        """
        Extracts NBA player statistics from Basketball Reference.
        """
        current_year = datetime.now().year
        url = f"https://www.basketball-reference.com/leagues/NBA_{current_year}_per_game.html"
        
        soup = self._safe_request(url)
        if soup is None:
            return pd.DataFrame()
        
        try:
            table = soup.find('table', {'id': 'per_game_stats'})
            if table:
                df = pd.read_html(str(table))[0]
                df = df[df['Rk'] != 'Rk']  # Remove repeated headers
                df['season'] = f"{current_year-1}-{current_year}"
                df['scraped_date'] = datetime.now().strftime('%Y-%m-%d')
                return df
        except Exception as e:
            print(f"Error: {e}")
        
        return pd.DataFrame()

# Create scraper instance
scraper = WebScraper()
print("[OK] WebScraper class loaded")
```

Use these APIs to create unique datasets by combining multiple data sources.


```python
class PublicAPIs:
    """
    Collection of free public APIs to generate unique datasets.
    No authentication required or have generous free tier.
    """
    
    def __init__(self):
        self.session = requests.Session()
    
    # REST Countries API - Country data
    def get_countries_data(self) -> pd.DataFrame:
        """Gets complete data for all countries in the world."""
        url = "https://restcountries.com/v3.1/all"
        try:
            response = self.session.get(url, timeout=30)
            data = response.json()
            
            countries = []
            for c in data:
                countries.append({
                    'name': c.get('name', {}).get('common', ''),
                    'official_name': c.get('name', {}).get('official', ''),
                    'capital': c.get('capital', [''])[0] if c.get('capital') else '',
                    'region': c.get('region', ''),
                    'subregion': c.get('subregion', ''),
                    'population': c.get('population', 0),
                    'area_km2': c.get('area', 0),
                    'languages': ', '.join(c.get('languages', {}).values()) if c.get('languages') else '',
                    'currencies': ', '.join([cur.get('name', '') for cur in c.get('currencies', {}).values()]) if c.get('currencies') else '',
                    'timezone': c.get('timezones', [''])[0] if c.get('timezones') else '',
                    'continent': c.get('continents', [''])[0] if c.get('continents') else '',
                    'landlocked': c.get('landlocked', False),
                    'borders': len(c.get('borders', [])),
                    'flag_emoji': c.get('flag', ''),
                    'gini_index': list(c.get('gini', {}).values())[0] if c.get('gini') else None,
                    'driving_side': c.get('car', {}).get('side', ''),
                    'independent': c.get('independent', True),
                    'un_member': c.get('unMember', False),
                })
            return pd.DataFrame(countries)
        except Exception as e:
            print(f"Error: {e}")
            return pd.DataFrame()
    
    # CoinGecko API - Cryptocurrencies
    def get_crypto_data(self, limit: int = 250) -> pd.DataFrame:
        """Gets data for top cryptocurrencies."""
        url = f"https://api.coingecko.com/api/v3/coins/markets"
        params = {
            'vs_currency': 'usd',
            'order': 'market_cap_desc',
            'per_page': limit,
            'page': 1,
            'sparkline': False,
            'price_change_percentage': '1h,24h,7d,30d'
        }
        try:
            response = self.session.get(url, params=params, timeout=30)
            data = response.json()
            df = pd.DataFrame(data)
            df['scraped_date'] = datetime.now().strftime('%Y-%m-%d %H:%M')
            return df
        except Exception as e:
            print(f"Error: {e}")
            return pd.DataFrame()
    
    # SpaceX API - Space launches
    def get_spacex_launches(self) -> pd.DataFrame:
        """Gets SpaceX launch history."""
        url = "https://api.spacexdata.com/v4/launches"
        try:
            response = self.session.get(url, timeout=30)
            data = response.json()
            
            launches = []
            for l in data:
                launches.append({
                    'name': l.get('name', ''),
                    'date_utc': l.get('date_utc', ''),
                    'success': l.get('success'),
                    'rocket': l.get('rocket', ''),
                    'details': l.get('details', '')[:200] if l.get('details') else '',
                    'flight_number': l.get('flight_number', 0),
                    'upcoming': l.get('upcoming', False),
                    'cores_reused': sum([1 for c in l.get('cores', []) if c.get('reused')]),
                    'fairings_recovered': l.get('fairings', {}).get('recovered') if l.get('fairings') else None,
                    'crew_count': len(l.get('crew', [])),
                    'payloads_count': len(l.get('payloads', [])),
                })
            return pd.DataFrame(launches)
        except Exception as e:
            print(f"Error: {e}")
            return pd.DataFrame()
    
    # Random User API - Fictional users
    def get_random_users(self, count: int = 1000) -> pd.DataFrame:
        """Generates realistic fictional users for datasets."""
        url = f"https://randomuser.me/api/?results={min(count, 5000)}&nat=us,gb,de,fr,es"
        try:
            response = self.session.get(url, timeout=60)
            data = response.json()
            
            users = []
            for u in data.get('results', []):
                users.append({
                    'first_name': u['name']['first'],
                    'last_name': u['name']['last'],
                    'gender': u['gender'],
                    'email': u['email'],
                    'age': u['dob']['age'],
                    'city': u['location']['city'],
                    'state': u['location']['state'],
                    'country': u['location']['country'],
                    'postcode': str(u['location']['postcode']),
                    'latitude': u['location']['coordinates']['latitude'],
                    'longitude': u['location']['coordinates']['longitude'],
                    'timezone_offset': u['location']['timezone']['offset'],
                    'registered_date': u['registered']['date'][:10],
                    'years_registered': u['registered']['age'],
                    'nationality': u['nat'],
                })
            return pd.DataFrame(users)
        except Exception as e:
            print(f"Error: {e}")
            return pd.DataFrame()
    
    # World Bank API - Economic indicators
    def get_world_bank_indicator(self, indicator: str = 'NY.GDP.PCAP.CD', year: int = 2022) -> pd.DataFrame:
        """
        Gets World Bank indicators.
        Useful indicators:
        - NY.GDP.PCAP.CD: GDP per capita
        - SP.POP.TOTL: Total population
        - SL.UEM.TOTL.ZS: Unemployment rate
        - SE.ADT.LITR.ZS: Literacy rate
        """
        url = f"https://api.worldbank.org/v2/country/all/indicator/{indicator}"
        params = {'format': 'json', 'date': year, 'per_page': 300}
        try:
            response = self.session.get(url, params=params, timeout=30)
            data = response.json()
            if len(data) > 1:
                df = pd.DataFrame(data[1])
                df = df[['country', 'value', 'date']]
                df['country_name'] = df['country'].apply(lambda x: x.get('value', '') if isinstance(x, dict) else '')
                df['country_code'] = df['country'].apply(lambda x: x.get('id', '') if isinstance(x, dict) else '')
                df = df[['country_name', 'country_code', 'value', 'date']]
                df.columns = ['country', 'country_code', indicator, 'year']
                return df.dropna()
            return pd.DataFrame()
        except Exception as e:
            print(f"Error: {e}")
            return pd.DataFrame()

# Create instance
apis = PublicAPIs()
print("[OK] PublicAPIs class loaded")
```
